```python
numeric_variable =  ['Age', 'BMI', 'AlcoholConsumption', 'PhysicalActivity','DietQuality', 'SleepQuality',
                   'SystolicBP', 'DiastolicBP', 'CholesterolTotal', 'CholesterolLDL', 'CholesterolHDL',
                   'CholesterolTriglycerides','UPDRS','MoCA','FunctionalAssessment']
# 轉換原始數據為長格式
melted_data = data.melt(id_vars=['Diagnosis'], value_vars=numeric_variable,
                        var_name='Feature', value_name='Value')

# 計算平均值、標準差和 t-test 的 p-value
table_data = []
for feature in numeric_variable:
    group_0 = melted_data.query("Feature == @feature & Diagnosis == 0")['Value']
    group_1 = melted_data.query("Feature == @feature & Diagnosis == 1")['Value']
    t_stat, p_value = ttest_ind(group_0, group_1, equal_var=False, nan_policy='omit')

    mean_std_0 = f"{group_0.mean():.1f} ± {group_0.std():.1f}"
    mean_std_1 = f"{group_1.mean():.1f} ± {group_1.std():.1f}"
    p_val_str = f"{p_value:.2f}" if p_value > 0.05 else f"*{p_value:.2f}*"

    table_data.append([feature, mean_std_0, mean_std_1, p_val_str])


# 繪製圖形
fig, (ax_box, ax_table) = plt.subplots(1, 2, figsize=(22, 10), gridspec_kw={'width_ratios': [1.3, 1]})

# 繪製箱型圖
sns.boxplot(data=melted_data, x='Feature', y='Value', hue='Diagnosis', palette='Set2', ax=ax_box)
ax_box.set_title('Boxplot of Selected Features by Diagnosis Groups', fontsize=16)
ax_box.set_xticklabels(ax_box.get_xticklabels(), rotation=45, ha='right', fontsize=14)
ax_box.legend(title='Diagnosis', loc='upper right')

# 添加表格
columns = ["Variable", "No-Parkinsons\n(N=801)", "Parkinsons\n(N=1304)", "p-value"]
ax_table.axis('off')  # 關閉表格軸
table = ax_table.table(cellText=table_data, colLabels=columns, loc='center', cellLoc='center', bbox=[0, 0, 1, 1])
table.auto_set_font_size(False)
table.set_fontsize(12)
table.scale(1.5, 2.5)

# 美化表格表頭和顯著性 p-value
for (row, col), cell in table.get_celld().items():
    if row == 0:  # 表頭行
        cell.set_text_props(weight='bold', color='white')
        cell.set_facecolor('#4CAF50')  # 綠色背景
    elif col == 3 and "*" in cell.get_text().get_text():  # 顯著性 p-value
        cell.set_text_props(color='red')

plt.tight_layout()
plt.show()
```

```python
import math
# 計算比例和卡方檢定的 p-value
def prepare_table_data(categorical_vars, data):
    table_data = []
    for feature in categorical_vars:
        contingency_table = pd.crosstab(data[feature], data['Diagnosis'])
        percentages_0 = (contingency_table[0] / contingency_table[0].sum() * 100).round(1)
        percentages_1 = (contingency_table[1] / contingency_table[1].sum() * 100).round(1)
        chi2, p_value, _, _ = chi2_contingency(contingency_table)
        p_val_str = f"{p_value:.2f}" if p_value > 0.05 else f"*{p_value:.2f}*"
        for i, category in enumerate(contingency_table.index):
            special_flag = (p_value < 0.05 and category == 1)  # 符合條件時設置標註
            table_data.append([
                f"{feature} - {category}",
                f"{percentages_0[category]}%",
                f"{percentages_1[category]}%",
                p_val_str if i == 0 else "",
                special_flag  # 將標註標誌加入數據
            ])
    return table_data

# 繪製表格
def plot_table(ax, table_data, columns):
    ax.axis('off')
    # 創建表格，忽略 special_flag 的數據
    table = ax.table(cellText=[row[:-1] for row in table_data], colLabels=columns, loc='center', cellLoc='center', bbox=[0, 0, 1, 1])
    table.auto_set_font_size(False)
    table.set_fontsize(14)
    table.scale(1.5, 2.5)
    for (row, col), cell in table.get_celld().items():
        if row == 0:
            cell.set_text_props(weight='bold', color='white')
            cell.set_facecolor('#4CAF50')
        elif col == 3 and "*" in cell.get_text().get_text():
            cell.set_text_props(color='red')
        # 特別標註 p-value < 0.05 且變數=1 的情況
        if row > 0 and table_data[row - 1][-1] and col in [1, 2]:  # 忽略標題列
            cell.set_text_props(weight='bold', color='blue')  # 標註變數=1 的結果

# 主要流程
categorical_variables = ['Gender', 'Ethnicity', 'EducationLevel', 'Smoking', 'FamilyHistoryParkinsons',
                         'TraumaticBrainInjury', 'Diabetes', 'Depression', 'Hypertension', 'Stroke',
                         'Tremor', 'Rigidity', 'Bradykinesia', 'PosturalInstability','SpeechProblems',
                         'SleepDisorders', 'Constipation']

# 假設 data 是你的數據
table_data = prepare_table_data(categorical_variables, data)
rows_per_table = math.ceil(20)

# 分割數據
table_data_1 = table_data[:rows_per_table]
table_data_2 = table_data[rows_per_table:]

# 創建左右佈局
fig = plt.figure(figsize=(26, 10))
gs = GridSpec(1, 2, width_ratios=[1, 1], wspace=0.05)

# 繪製表格
columns = ["Variable", "No-Parkinsons (%)", "Parkinsons (%)", "p-value"]
plot_table(fig.add_subplot(gs[0]), table_data_1, columns)
plot_table(fig.add_subplot(gs[1]), table_data_2, columns)

plt.subplots_adjust(left=0.05, right=0.95, top=0.95, bottom=0.05, wspace=0.1)
plt.show()
```

```python
X1 = data[['UPDRS', 'Tremor', 'FunctionalAssessment', 'Rigidity', 'Bradykinesia',
           'MoCA', 'PosturalInstability', 'Age', 'Depression', 'Diabetes','Gender']]
y1 = data['Diagnosis']

#split data into training and testing sets
X1_train, X1_test, y1_train, y1_test = train_test_split(X1, y1, test_size = 0.2, random_state = 921, shuffle = True)
```

<div style="margin: 10px auto; max-width: 100%; text-align: left; font-family: Palatino Linotype; font-size: 1.4em; padding: 5px 5px; background:#ffffff; border-radius: 10px;">
    <strong>IV-4-2. Standardization and OneHotEncoder </strong>
</div>

```python
# Define numerical and categorical columns
numerical1 = ['Age','FunctionalAssessment','MoCA','UPDRS']

categorical1 = ['Gender','Rigidity','Bradykinesia','Tremor','PosturalInstability', 'Depression','Diabetes']

# Step 1: Standardize numerical columns
standard_scaler = StandardScaler()
X1_train[numerical1] = standard_scaler.fit_transform(X1_train[numerical1])  # Fit and transform on training data
X1_test[numerical1] = standard_scaler.transform(X1_test[numerical1])  # Transform only on test data

# Step 2: One-hot encode categorical columns
ohe_encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')  # Avoid errors for unseen categories
X1_train_ohe = ohe_encoder.fit_transform(X1_train[categorical1])  # Fit and transform on training data
X1_test_ohe = ohe_encoder.transform(X1_test[categorical1])  # Transform only on test data

# Step 3: Convert encoded data to DataFrames
X1_train_ohe_df = pd.DataFrame(X1_train_ohe, columns=ohe_encoder.get_feature_names_out(categorical1), index=X1_train.index)
X1_test_ohe_df = pd.DataFrame(X1_test_ohe, columns=ohe_encoder.get_feature_names_out(categorical1), index=X1_test.index)

# Step 4: Drop original categorical columns and concatenate encoded columns
X1_train = pd.concat([X1_train[numerical1], X1_train_ohe_df], axis=1)
X1_test = pd.concat([X1_test[numerical1], X1_test_ohe_df], axis=1)
```

<div style="margin: 10px auto; max-width: 100%; text-align: left; font-family: Palatino Linotype; font-size: 1.4em; padding: 5px 5px; background:#ffffff; border-radius: 10px;">
    <strong>IV-4-3. Machine Learning Models and Tuning Hyperparameter </strong>
</div>

```python
#define hyperparameter grids for each model
param_grids = {
    'Decision Tree': {'max_depth': [None,3, 5],'criterion':['gini','entropy'],'max_leaf_nodes': [None, 10, 20, 50],'class_weight': ['balanced']},
    'Random Forest': {'n_estimators': [50, 100, 200], 'max_depth': [None,3, 7, 12],'class_weight': ['balanced']},
    'K-Nearest Neighbors': {'n_neighbors': [10, 15, 20],'leaf_size': [3, 5, 10]},
    'Logistic Regression': {'C': [0.01,0.1, 1],'class_weight': ['balanced'],'max_iter': [100, 200, 500]},
    'Support Vector Machine': {'C': [0.1, 1, 10], 'gamma': [0.1, 1, 'scale', 'auto'],'class_weight': ['balanced']},
    'XGBoost': {'n_estimators': [50, 100, 200], 'learning_rate': [0.01, 0.1, 1], 'max_depth': [3, 5, 7]},
    }

#instantiate classification models with default parameters
models = {
    'Decision Tree': DecisionTreeClassifier(),
    'Random Forest': RandomForestClassifier(),
    'K-Nearest Neighbors': KNeighborsClassifier(),
    'Logistic Regression': LogisticRegression(),
    'Support Vector Machine': SVC(),
    'XGBoost': XGBClassifier(),
}
```

<div style="margin: 10px auto; max-width: 100%; text-align: left; font-family: Palatino Linotype; font-size: 1.4em; padding: 5px 5px; background:#ffffff; border-radius: 10px;">
    <strong>IV-4-4. Evaluation of Machine Learning Models </strong>
</div>

```python
# Define a DataFrame to store the metrics
metrics_data1 = []

# Fit models using GridSearchCV for hyperparameter tuning and collect metrics
for name, model in models.items():
    grid_search = GridSearchCV(model, param_grids[name], cv=5, scoring='accuracy')
    grid_search.fit(X1_train, y1_train)
    best_model = grid_search.best_estimator_
    y1_pred = best_model.predict(X1_test)

    # Extract metrics
    accuracy = accuracy_score(y1_test, y1_pred)
    precision = precision_score(y1_test, y1_pred, average='weighted', zero_division=0)
    recall = recall_score(y1_test, y1_pred, average='weighted', zero_division=0)
    f1 = f1_score(y1_test, y1_pred, average='weighted', zero_division=0)

    # Append metrics to the DataFrame
    metrics_data1.append({'Model': name, 'Accuracy': accuracy, 'Precision': precision, 'Recall': recall, 'F1 Score': f1})

    # Print classification report and best parameters
    report = classification_report(y1_test, y1_pred)
    print(f'{name} Classification Report:\n{report}\nBest Parameters: {grid_search.best_params_}\n')
```
